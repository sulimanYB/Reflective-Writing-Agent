
# Reflective Writing Agent
### An Agentic AI Workflow for Iterative Essay Improvement

## 1. Introduction

Large Language Models (LLMs) can generate coherent written content,
but their initial responses may contain weaknesses in structure,
clarity, argumentation and writing style.

Reflection is an agentic AI design pattern in which an LLM
critically evaluates an initial output and uses the resulting
feedback to improve a subsequent version.

This project demonstrates a four-stage workflow:

1. **Draft Generation:** Generate an initial essay from a user-defined topic.
2. **Reflection:** Critically assess the essay and identify weaknesses.
3. **Revision:** Improve the original essay using the reflection feedback.
4. **Evaluation:** Compare the original and revised essays using an LLM as a judge.

### Project Objectives

- Demonstrate the reflection design pattern in agentic AI.
- Implement a modular workflow using the OpenAI API.
- Apply LLM-based evaluation to compare writing quality.
- Visualise changes in essay quality across four evaluation criteria.

The workflow is implemented in Python, with Jupyter Notebook
providing an interactive environment for experimentation,
analysis and visualisation.



## 2. Workflow Architecture

The Reflective Writing Agent implements a sequential, four-stage
LLM workflow based on the **Reflection design pattern**.

Each stage has a distinct responsibility, and its output is
passed to the next stage.

### Stage 1 — Draft Generation
The Drafting Agent receives an essay topic and generates an
initial structured essay using an OpenAI language model.

### Stage 2 — Reflection
The Reflection Agent examines the initial draft and provides
constructive feedback on structure, clarity, argument strength
and writing style.

### Stage 3 — Revision
The Revision Agent receives both the original draft and the
reflection feedback. It produces an improved essay that
addresses the identified weaknesses.

### Stage 4 — Evaluation
An independent evaluation step compares the original and
revised essays using an LLM-as-a-Judge approach.

Both essays receive scores from 1 to 10 across four criteria:
structure, clarity, arguments and writing style.

The resulting scores are used to visualise changes in
LLM-assessed writing quality.

### Architecture Summary

**Essay Topic**
→ **Drafting**
→ **Reflection**
→ **Revision**
→ **Evaluation**
→ **Results and Visualisation**

The Evaluation Agent compares both the original and revised
essays rather than evaluating the revised essay in isolation.

This implementation performs a single reflection–revision
cycle. It does not automatically repeat the cycle or use
evaluation scores to trigger additional revisions.


## 4. Draft Generation

The first stage of the workflow generates an initial essay
based on a user-defined topic.

The **Drafting Agent** uses an OpenAI language model to produce
a structured response containing an introduction, supporting
arguments and a conclusion.

### How It Works

1. The user provides an essay topic.
2. The topic is incorporated into a prompt containing
   instructions for essay generation.
3. The prompt is submitted to the language model.
4. The generated essay is returned as a string.

### Implementation

The `generate_draft()` function, defined in `agents.py`,
performs the LLM request.

The generated essay is stored in the `draft` variable and
displayed using Jupyter's Markdown rendering.

### Purpose

This initial draft establishes the baseline for the
subsequent reflection, revision and evaluation stages.

At this stage, the essay has not undergone any automated
critical review or refinement.


```python
from IPython.display import display, Markdown

# Define the essay topic
topic = "Should social media platforms be regulated by the government?"

# Step 1 — Generate the initial draft
draft = generate_draft(topic)

# Display the essay as formatted Markdown
display(Markdown("## Step 1 — Initial Essay Draft"))
display(Markdown(draft))
```


## 5. Reflection

The second stage implements the **Reflection design pattern**,
a technique in agentic AI where an LLM critically examines
previously generated content to identify weaknesses and
opportunities for improvement.

Rather than immediately accepting the initial essay, the
workflow introduces a dedicated review stage.

### How It Works

1. The initial essay is passed to the Reflection Agent.
2. The LLM evaluates the essay against four criteria:
   - Structure and organisation
   - Clarity and coherence
   - Strength of arguments and supporting evidence
   - Writing style and readability
3. The model identifies weaknesses and recommends improvements.
4. The resulting feedback is returned as a string.

### Implementation

The `reflect_on_draft()` function, defined in `agents.py`,
performs the critical review.

The feedback is stored in the `feedback` variable and
displayed using Jupyter's Markdown rendering.

### Purpose

The Reflection Agent separates content generation from
critical assessment, allowing the workflow to identify
potential improvements before producing a revised essay.

Importantly, this stage generates **feedback only**.
It does not rewrite the essay.

The effectiveness of the reflection depends on the quality
of the model's critique, which may itself contain errors,
omissions or subjective judgements.


```python
# Step 2 — Reflect on the initial draft
feedback = reflect_on_draft(draft)

# Display the feedback as formatted Markdown
display(Markdown("## Step 2 — Reflection and Feedback"))
display(Markdown(feedback))
```


## 6. Revision

The third stage applies the feedback generated during reflection
to produce an improved version of the original essay.

The **Revision Agent** receives both the initial draft and
the constructive feedback from the Reflection Agent.

### How It Works

1. The original essay and reflection feedback are provided
   as inputs to the Revision Agent.
2. The LLM is instructed to address the weaknesses identified
   during reflection.
3. The model generates a revised essay while preserving
   the original topic and main ideas.
4. The revised essay is returned as a string.

### Implementation

The `revise_draft()` function, defined in `agents.py`,
combines the original draft and reflection feedback
into a revision prompt.

The resulting essay is stored in the `revised` variable
and displayed using Jupyter's Markdown rendering.

### Purpose

This stage demonstrates how feedback from one LLM call
can guide a subsequent generation step.

The revision aims to improve:

- Structure and organisation
- Clarity and coherence
- Strength of arguments
- Writing style and overall flow

### Expected Outcome

The revised essay should address the weaknesses identified
during reflection while retaining the original meaning.

However, revision does not guarantee improvement.
The effectiveness of the changes must be assessed separately,
which motivates the evaluation stage that follows.


```python
# Step 3 — Revise the essay using the reflection feedback
revised = revise_draft(draft, feedback)

# Display the revised essay as formatted Markdown
display(Markdown("## Step 3 — Revised Essay"))
display(Markdown(revised))
```

## 7. Evaluation Using LLM-as-a-Judge

The final stage evaluates whether the reflection and revision process has improved the quality of the original essay.

This stage uses an **LLM-as-a-Judge** approach, in which a language model assesses generated content against predefined evaluation criteria.

### Evaluation Criteria

Both essays are evaluated using four criteria:

1. **Structure:** Logical organisation, paragraph structure and progression of ideas.
2. **Clarity:** Readability, coherence and clarity of expression.
3. **Arguments:** Strength of reasoning, supporting evidence and consideration of alternative perspectives.
4. **Writing Style:** Grammar, vocabulary, tone and overall presentation.

Each criterion receives a score from **1 to 10** for both the original and revised essays.

### How It Works

1. The original and revised essays are passed to the Evaluation Agent.
2. The LLM compares both versions using the same evaluation criteria.
3. The model assigns numerical scores and provides an overall assessment.
4. The results are returned in JSON format.
5. The JSON response is converted into a Python dictionary and subsequently organised into a Pandas DataFrame.

### Implementation

The `evaluate_essays()` function, defined in `agents.py`, performs the comparison.

The evaluation uses `gpt-4o-mini` with a temperature of `0.2` to encourage more consistent responses.

The model is explicitly instructed **not to assume that the revised essay is better**.

### Evaluation Limitations

LLM-as-a-Judge provides a scalable and convenient method for evaluating generated content, but its scores are inherently subjective.

Several limitations should be considered:

- The evaluator may exhibit biases related to wording, presentation or response length.
- Scores may vary between repeated evaluations.
- Using the same model family for generation and evaluation does not guarantee independent assessment.
- A higher score does not necessarily establish objectively superior writing quality.

Consequently, the evaluation scores should be interpreted as **LLM-assessed writing quality**, rather than independently validated performance measurements.

The evaluation provides an initial indication of whether the reflection process has produced meaningful improvements.

```python
import json

# Step 4 — Evaluate the original and revised essays
evaluation = evaluate_essays(draft, revised)

# Convert the JSON response into a Python dictionary
scores = json.loads(evaluation)

# Display the evaluation results
display(Markdown("## Step 4 — Essay Evaluation"))
display(Markdown(f"**Overall Assessment:** {scores['overall_assessment']}"))

# Display the evaluation scores
import pandas as pd

criteria = ["structure", "clarity", "arguments", "writing_style"]

df = pd.DataFrame({
    "Criterion": [c.replace("_", " ").title() for c in criteria],
    "Original Draft": [scores[c]["original"] for c in criteria],
    "Revised Draft": [scores[c]["revised"] for c in criteria]
})

display(df)
```

## 8. Results and Visualisation

The evaluation results are presented using a comparative bar chart to illustrate changes in LLM-assessed writing quality following the reflection and revision process.

### Evaluation Results

The Evaluation Agent assigned scores to both essays across four criteria: structure, clarity, argument strength and writing style.

The results are stored in a Pandas DataFrame and visualised using Matplotlib.

### Visualisation Approach

A horizontal grouped bar chart is used to compare the original and revised essays.

- **Grey bars:** Original essay scores.
- **Teal bars:** Revised essay scores.
- **Horizontal axis:** Evaluation scores ranging from 1 to 10.
- **Vertical axis:** Writing quality criteria.

Numerical labels are displayed alongside each bar to facilitate comparison.

### Interpretation

The chart provides a visual indication of how the revised essay performs relative to the original across the four evaluation criteria.

Positive score differences suggest that the evaluator judged the revision favourably, while negative differences would indicate a potential deterioration.

These results should be interpreted cautiously because they are derived from a single LLM evaluation rather than repeated independent assessments.

```python
import numpy as np
import matplotlib.pyplot as plt

# Extract evaluation data
labels = df["Criterion"]
original = df["Original Draft"]
revised = df["Revised Draft"]

y = np.arange(len(labels))
height = 0.32

# Create figure
fig, ax = plt.subplots(figsize=(10, 5.5))

# Horizontal comparison bars
bars_original = ax.barh(
    y + height / 2, original, height,
    label="Original Draft", color="#94A3B8"
)

bars_revised = ax.barh(
    y - height / 2, revised, height,
    label="Revised Draft", color="#0D9488"
)

# Add score labels
ax.bar_label(bars_original, padding=4, fontsize=11)
ax.bar_label(bars_revised, padding=4, fontsize=11)

# Chart formatting
ax.set_yticks(y)
ax.set_yticklabels(labels, fontsize=12)
ax.invert_yaxis()

ax.set_xlim(0, 10.7)
ax.set_xlabel("LLM Evaluation Score (1–10)", fontsize=11)

ax.set_title(
    "Reflective Writing Agent\nOriginal vs Revised Essay",
    fontsize=17, fontweight="bold", pad=20
)

ax.legend(loc="lower right", frameon=False)

# Remove unnecessary borders
for spine in ax.spines.values():
    spine.set_visible(False)

ax.grid(axis="x", alpha=0.15)
ax.set_axisbelow(True)

plt.tight_layout()

# Save the evaluation chart for GitHub
fig.savefig(
    "images/essay_evaluation.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```

## 9. Discussion and Limitations

### 9.1 Discussion

This experiment demonstrates the application of the **Reflection design pattern** to an automated writing task.

The workflow separates essay generation, critical review, revision and evaluation into four distinct LLM-based stages.

In the example examined, the Evaluation Agent assigned an average score of **7.5/10** to the original essay and **8.5/10** to the revised essay, representing an improvement of **1.0 point**.

All four evaluation criteria showed an increase of one point following revision.

These findings suggest that the reflection and revision process produced an essay that the evaluator considered better structured, clearer and more effective.

However, the experiment does not establish that reflective writing consistently improves essay quality across different topics, models or evaluation settings.

### 9.2 Limitations

**Single Evaluation**

The experiment examines one essay topic and a single reflection–revision cycle. The results cannot be generalised to other writing tasks without additional experiments.

**LLM-as-a-Judge Bias**

The evaluation relies on subjective judgements generated by an LLM. Scores may be affected by the wording of the evaluation prompt, response length and model-specific preferences.

**Lack of Independent Validation**

The same model is used across the writing and evaluation stages. Independent human assessments or evaluations using different models would provide stronger evidence of improvement.

**No Iterative Refinement**

The current workflow performs only one revision cycle. It does not use evaluation feedback to initiate further revisions when the output fails to meet a predefined quality threshold.

**Factual Accuracy**

The evaluation focuses primarily on writing quality. It does not independently verify factual claims, citations or supporting evidence within the essays.

### 9.3 Potential Improvements

Future extensions could include:

- Implementing multiple reflection–revision cycles with a stopping criterion.
- Comparing different language models for drafting, reflection and evaluation.
- Evaluating performance across a larger dataset of essay topics.
- Introducing human evaluation to validate LLM-generated scores.
- Adding factual verification and source-grounded research.
- Measuring API costs, token usage and execution time.

These extensions would allow a more rigorous assessment of the effectiveness, reliability and efficiency of reflection-based LLM workflows.
